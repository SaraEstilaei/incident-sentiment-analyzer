```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re
from sklearn.model_selection import train_test_split
train_path = r"C:\Users\Ramtin\Desktop\uni\quera Ai\deep learning\nlp\train_data.csv"
test_path = r"C:\Users\Ramtin\Desktop\uni\quera Ai\deep learning\nlp\test_data.csv"
title_brand_path = r"C:\Users\Ramtin\Desktop\uni\quera Ai\deep learning\nlp\title_brand.csv"
df = pd.read_csv(train_path)  

print("SHAPE:")
print(df.shape)

print("COLUMNS:")
print(list(df.columns))

print("INFO:")
display(df.info())

display(df.head(15))

print("NULL COUNTS:")
display(df.isnull().sum())
```

```python
df['vote'] = df['vote'].astype(str).str.replace(',', '').replace('nan', '0')
df['vote'] = pd.to_numeric(df['vote'], errors='coerce').fillna(0).astype(int)
df['summary'] = df['summary'].fillna('')
df['reviewTime'] = pd.to_datetime(df['reviewTime'])
df['review_year'] = df['reviewTime'].dt.year
print(df.info())
```

```python
def clean_text_light(text):
    text = re.sub(r'<.*?>', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['reviewText_clean'] = df['reviewText'].apply(clean_text_light)

df['full_text'] = df['summary'] + " " + df['reviewText_clean']
df['full_text'] = df['full_text'].str.strip()
sample, _ = train_test_split(df, train_size=5000, stratify=df['overall'], random_state=42)
train_subset, _ = train_test_split(df, train_size=50000, stratify=df['overall'], random_state=42)
sample.to_csv('sample_5k.csv', index=False)
train_subset.to_csv('Train_subset_50k.csv', index=False)
df.to_csv('df_cleaned.csv', index=False)
```

```python
df_test = pd.read_csv(test_path)
display(df_test.head())
df_test['vote'] = df_test['vote'].astype(str).str.replace(',', '').replace('nan', '0')
df_test['vote'] = pd.to_numeric(df_test['vote'], errors='coerce').fillna(0).astype(int)
df_test['summary'] = df_test['summary'].fillna('')
df_test['reviewTime'] = pd.to_datetime(df_test['reviewTime'])
df_test['review_year'] = df_test['reviewTime'].dt.year
df_test['verified'] = df_test['verified'].astype(int)
df_test['reviewText_clean'] = df_test['reviewText'].apply(clean_text_light)
df_test['full_text'] = (df_test['summary'] + " " + df_test['reviewText_clean']).str.strip()
df_test.to_csv('test_cleaned.csv', index=False)
display(df_test.head())
```
